```python
# Print numbers 1–20 using a for loop
for i in range(1,21):
    print(i)
```

```python
# Print even numbers from 2–20

for i in range(2,21):
    if i%2==0:
        print(i)
        
```

```python
# Print odd numbers from 1–19
for i in range(1,20):
    if i%2!=0:
        print(i)
    
```

```python
# Print the table of any number entered by the user
num=int(input("enter the number "))
for i in range(1,11):
    print(f"{num} x {i} = {num*i}")
    print(i)
```

```python
# Find the sum of numbers from 1–100
total=0
for i in range(1,101):
    total+=i
print(total)
```

```python
# Count down from 10 to 1
for i in range(10,0,-1):
    print(i)
```

```python
# Print every character of your name using a loop
name = input("Enter your name: ")

for letter in name:
    print(letter)
```

```python
# Use a while loop to print 1–10
i=1
while i<=10:
    print(i)
    i+=1
```

```python
# Use break to stop a loop when the number becomes 7
for i in range(1,11):
    if i==7:
        break
    print(i)
```

```python
# Use continue to skip the number 5
for i in range(1,11):
    if i==5:
        continue
    
    print(i)
```

```python
# Print numbers 1 to 10 with square

for i in range(1,11):
    print(i, "square =", i*i )
    
```

```python
# Count numbers from 1 to 100 which are divisible by 5
count=0
for i in range(1,101):
    if i%5==0:
     count+=1
    
print("TOTAL NUMBER",count)
```

```python
# Find largest number in a list
numbers=[10,20,30,40,50]
largest=numbers[0]
for num in numbers:
    if num > largest:
        largest=num
        
print("Largest number ",largest)
    
```

```python
# Find Second largest number in a list
numbers=[10,20,30,40,50]
largest=numbers[0]
second_largset=numbers[0]
for num in numbers:
    if num>largest:
        second_largset=largest
        largest=num
        
    elif num>second_largset and num!=largest:
        second_largset=num
        
print('LARGEST NUMBER',largest)
print('SECOND LARGSET',second_largset)
```

```python
# Reverse a string using loop
name='PYTHON'
reverse=""
for char in name:
    reverse=char+reverse
print(reverse)
```

```python
# Count vowels in a word
word='programing'
count=0
for char in word:
    if char in 'aeiou':
        count+=1
print("vowels",count)
    


    
    
```
